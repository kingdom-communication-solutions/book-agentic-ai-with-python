## Refine the Instructions

A useful workshop plan should reflect the experience of the facilitator and the needs of the participants. Instead of asking the model to simply create a plan, we can instruct it to approach the task as an experienced workshop facilitator.

Let's change our instructions:

```python
system_prompt = """You are an experienced workshop facilitator
specialising in introductory AI education.
Design engaging and practical workshops
for beginners."""
```

We leave the user prompt unchanged and run the program again.

```python
response = get_completion(
    system_prompt,
    user_prompt,
    model="YOUR_MODEL_NAME"
)
print(response)
```

Predict how the response might differ before running the code.

Assigning a role gives the model a more specific perspective from which to approach the task. However, a role alone is not enough. The assistant also needs information about the workshop and the conditions under which it must be organised.

Let's refine our instructions further:

```python
system_prompt = """You are an experienced workshop facilitator
specialising in introductory AI education.
Design engaging and practical workshops
for beginners.
The participants have no prior AI knowledge.
They have laptops and internet access.
Use free tools only.
Combine short explanations with
hands-on activities."""
```

Predict what changes after adding context and constraints, then run the request.

```python
response = get_completion(
    system_prompt,
    user_prompt,
    model="YOUR_MODEL_NAME"
)
print(response)
```

We have now introduced context and constraints. The assistant knows who the workshop is for, what resources are available and how the learning experience should be organised.

## Structure the Output

A workshop programme is easier to use when its information is organised clearly. Instead of receiving an unstructured response, we want a programme that allows us to quickly identify the time, topic, activity and learning outcome of each session.

We therefore extend our instructions to define the output format and duration.

```python
system_prompt = """You are an experienced workshop facilitator
specialising in introductory AI education.
Design engaging and practical workshops
for beginners.
The participants have no prior AI knowledge.
They have laptops and internet access.
Use free tools only.
Combine short explanations with
hands-on activities.
The workshop must last exactly 120 minutes.
Present the plan as a Markdown table
with these columns:
Time | Topic | Activity | Learning Outcome
Ensure the time allocations add up
to exactly 120 minutes."""
```

We can now generate the improved workshop programme using our existing function.

```python
response = get_completion(
    system_prompt,
    user_prompt,
    model="YOUR_MODEL_NAME"
)
print(response)
```

## Evaluate and Improve the Result

Generating a workshop plan is only part of the process. We must determine whether the result is actually suitable for our intended audience.

Consider the generated programme. Does it offer enough practical experience? Are the learning outcomes clear? Are the activities realistic within the allocated time?

We can evaluate the result using a few simple criteria.

[Table: Criteria for evaluating the workshop programme]

| Criterion | What to check |
| --- | --- |
| Duration | Does the programme total 120 minutes? |
| Audience | Is it suitable for beginners? |
| Activities | Can participants realistically complete them? |
| Resources | Are the required tools accessible and free? |
| Learning outcomes | Does each session contribute to the workshop's goal? |

If the programme does not meet our expectations, we can refine the instructions further. For example, we might require participants to complete a practical task using an AI tool rather than merely observing a demonstration.

This process illustrates an important principle: AI-generated content is a starting point for our work, not a substitute for our judgment.
